## 4. Broadcasting Operations

### Creating a 3x3 array with random integers and a 1D array with three elements

```python
array_3x3 = np.random.randint(1, 10, size=(3, 3))
array_1d_3elements = np.random.randint(1, 10, size=(3))

print("3x3 array (array_3x3):")
display(array_3x3)

print("1D array with 3 elements (array_1d_3elements):")
display(array_1d_3elements)
```

### Performing Addition using Broadcasting

```python
addition_result = array_3x3 + array_1d_3elements
print("Result of addition (array_3x3 + array_1d_3elements):")
display(addition_result)
```

#### Explanation of Broadcasting for Addition:

When `array_3x3` (shape: `(3, 3)`) is added to `array_1d_3elements` (shape: `(3,)`), NumPy applies broadcasting rules.

1.  **Alignment**: The dimensions are compared starting from the trailing (rightmost) dimension.
    -   `array_3x3`: `(3, 3)`
    -   `array_1d_3elements`: `(3,)` becomes `(1, 3)` internally for alignment purposes.

2.  **Compatibility**: For each dimension, the sizes must either be equal, or one of them must be 1. If one dimension is 1, it's 'stretched' or 'broadcast' to match the other.
    -   Rightmost dimension: `3` vs `3` (equal, compatible).
    -   Next dimension (left): `3` vs `1` (one is 1, compatible). The `1` is stretched to `3`.

Effectively, the `array_1d_3elements` is treated as if it were a 2D array of shape `(3, 3)` where each row is a copy of the original 1D array. Then, element-wise addition is performed.

### Performing Subtraction using Broadcasting

```python
subtraction_result = array_3x3 - array_1d_3elements
print("Result of subtraction (array_3x3 - array_1d_3elements):")
display(subtraction_result)
```

#### Explanation of Broadcasting for Subtraction:

The broadcasting mechanism for subtraction is identical to that for addition. The `array_1d_3elements` (shape `(3,)`) is broadcast across the rows of `array_3x3` (shape `(3, 3)`). This means each element of `array_1d_3elements` is subtracted from the corresponding column elements in each row of `array_3x3`.

### Performing Multiplication using Broadcasting

```python
multiplication_result = array_3x3 * array_1d_3elements
print("Result of multiplication (array_3x3 * array_1d_3elements):")
display(multiplication_result)
```

#### Explanation of Broadcasting for Multiplication:

Similar to addition and subtraction, element-wise multiplication leverages the same broadcasting rules. The `array_1d_3elements` is conceptually stretched to match the shape of `array_3x3` by replicating its values across the rows. Then, corresponding elements are multiplied together.

## 5. Matrix Operations

### Creating two 3x3 matrices with random integers

```python
matrix_a = np.random.randint(1, 10, size=(3, 3))
matrix_b = np.random.randint(1, 10, size=(3, 3))

print("Matrix A:")
display(matrix_a)

print("Matrix B:")
display(matrix_b)
```

### Performing Matrix Multiplication (Dot Product)

```python
matrix_multiplication_result = np.dot(matrix_a, matrix_b)
print("Result of Matrix Multiplication (A @ B):")
display(matrix_multiplication_result)
```

### Performing Element-wise Multiplication

```python
elementwise_multiplication_result = matrix_a * matrix_b
print("Result of Element-wise Multiplication (A * B):")
display(elementwise_multiplication_result)
```

### Computing the Determinant of one matrix (Matrix A)

```python
determinant_a = np.linalg.det(matrix_a)
print("Determinant of Matrix A:")
display(determinant_a)
```

### Finding the Eigenvalues and Eigenvectors of one matrix (Matrix A)

```python
eigenvalues, eigenvectors = np.linalg.eig(matrix_a)

print("Eigenvalues of Matrix A:")
display(eigenvalues)

print("Eigenvectors of Matrix A:")
display(eigenvectors)
```

## 6. Statistical Operations and Normalization

### Creating a 2D array of shape 5x4 with random floating-point numbers between 0 and 1

```python
random_float_array = np.random.rand(5, 4)
print("Original 5x4 array of random floats:")
display(random_float_array)
```

### Calculating the mean, median, and standard deviation of the entire array

```python
array_mean = np.mean(random_float_array)
array_median = np.median(random_float_array)
array_std = np.std(random_float_array)

print(f"Mean of the array: {array_mean:.4f}")
print(f"Median of the array: {array_median:.4f}")
print(f"Standard Deviation of the array: {array_std:.4f}")
```

### Finding the maximum and minimum values along each row

```python
max_per_row = np.max(random_float_array, axis=1)
min_per_row = np.min(random_float_array, axis=1)

print("Maximum value per row:")
display(max_per_row)

print("Minimum value per row:")
display(min_per_row)
```

### Normalizing the array (scaling values between 0 and 1)

```python
min_val = np.min(random_float_array)
max_val = np.max(random_float_array)

normalized_array = (random_float_array - min_val) / (max_val - min_val)

print("Normalized array (scaled between 0 and 1):")
display(normalized_array)
```

## 7. Sorting and Searching

### Creating a 1D array with random integers between 1 and 100

```python
random_1d_array = np.random.randint(1, 101, size=15) # Create an array of 15 random integers
print("Original 1D array:")
display(random_1d_array)
```

### Sorting the array in ascending order

```python
sorted_array = np.sort(random_1d_array)
print("Sorted array (ascending order):")
display(sorted_array)
```

### Finding the indices of the sorted elements

```python
sorted_indices = np.argsort(random_1d_array)
print("Indices that would sort the array:")
display(sorted_indices)
print("Original array reordered using sorted_indices:")
display(random_1d_array[sorted_indices]) # Verify it matches sorted_array
```

### Searching for a specific number

```python
search_number = 50 # Example number to search for

# Using np.where to find indices of the number
indices_where = np.where(random_1d_array == search_number)
print(f"Indices where '{search_number}' is found (using np.where): {indices_where}")

# Demonstrating np.argmin (finds the index of the minimum value)
# To search for a specific value, we can find the minimum absolute difference
# This is useful when the exact value might not exist, or to find the closest.
diff_array = np.abs(random_1d_array - search_number)
index_argmin = np.argmin(diff_array)
closest_value = random_1d_array[index_argmin]

print(f"The value closest to '{search_number}' is '{closest_value}' at index '{index_argmin}' (using np.argmin for closest value).")
```

## 10. Solving Systems of Linear Equations

### Define the coefficient matrix (A) and the constant vector (B)

```python
A = np.array([[2, 1, -1],
              [-3, -1, 2],
              [-2, 1, 2]])

B = np.array([8, -11, -3])

print("Coefficient Matrix A:")
display(A)

print("Constant Vector B:")
display(B)
```

### Solve the system of linear equations using `np.linalg.solve`

```python
X = np.linalg.solve(A, B)
print("Solution X (x, y, z):")
display(X)
```

### Verify the solution by substituting it back into the equations

```python
verification = np.dot(A, X)
print("Verification (A @ X):")
display(verification)

print("Does A @ X approximately equal B?")
display(np.allclose(verification, B))
```

## 11. Generating Structured Data

## 13. Performance Comparison: NumPy vs Python Loops

### Create a 1000x1000 array of random integers between 1 and 10.

```python
large_array = np.random.randint(1, 11, size=(1000, 1000))
print("Large 1000x1000 array created (displaying first 5x5 for brevity):")
display(large_array[:5, :5])
```

### Compute the sum of all elements using a Python loop and measure execution time.

```python
%%timeit

python_loop_sum = 0
for row in large_array:
    for element in row:
        python_loop_sum += element
```

### Compute the sum of all elements using NumPy's `sum` function and measure execution time.

```python
%%timeit

numpy_sum = np.sum(large_array)
```

### Compare the execution times.

```python
### Comparison of Execution Times using `%%timeit`

`%%timeit` runs the code multiple times and reports the average execution time along with standard deviation, providing a more accurate benchmark than a single `%%time` or manual `time.time()` measurement. The outputs above from each cell will now reflect this more precise timing.
```

From the `%%timeit` results:

*   **Python loop**: approximately 108 ms
*   **NumPy's `sum`**: approximately 529 µs (0.529 ms)

This shows that NumPy's `sum` function is significantly faster, approximately `108 ms / 0.529 ms ≈ 204` times faster than the Python loop for summing elements in a 1000x1000 array. This highlights the substantial performance benefits of using vectorized operations in NumPy over traditional Python loops for numerical computations.

## 14. Manipulating Array Shapes

### Creating a 3D array of shape 3x3x3 with random integers.

```python
array_3d = np.random.randint(1, 28, size=(3, 3, 3))
print("Original 3D array (3x3x3):")
display(array_3d)
```

### Reshaping it into a 1D array and then back into a 2D array of shape 9x3.

```python
# Reshape to 1D array
reshaped_1d = array_3d.reshape(-1)
print("Reshaped to 1D array:")
display(reshaped_1d)

# Reshape to 2D array of shape 9x3
reshaped_2d_9x3 = reshaped_1d.reshape(9, 3)
print("Reshaped to 2D array (9x3):")
display(reshaped_2d_9x3)
```

### Swap the axes of the original 3D array.

```python
# Original shape is (depth, rows, columns)
# Swapping axes to (rows, depth, columns) for example
swapped_axes_array = np.swapaxes(array_3d, 0, 1) # Swapping depth (axis 0) with rows (axis 1)
print("Original 3D array shape:", array_3d.shape)
print("Swapped axes array shape (axis 0 and 1 swapped):", swapped_axes_array.shape)
print("Swapped axes array (first 2 layers for brevity):")
display(swapped_axes_array[0:2, :, :])

# You can also use .transpose() for more complex permutations
# For example, to change (0, 1, 2) to (2, 0, 1)
permuted_axes_array = array_3d.transpose((2, 0, 1))
print("Permuted axes array shape (axes 0,1,2 -> 2,0,1):", permuted_axes_array.shape)
print("Permuted axes array (first 2 layers for brevity):")
display(permuted_axes_array[0:2, :, :])
```

### Split the reshaped 2D array into three equal parts along rows.

```python
split_arrays = np.array_split(reshaped_2d_9x3, 3, axis=0)

print("Original 2D array (9x3):")
display(reshaped_2d_9x3)

print("First part (3x3):")
display(split_arrays[0])

print("Second part (3x3):")
display(split_arrays[1])

print("Third part (3x3):")
display(split_arrays[2])
```
